# **Day7-First Chart Story : Turning Numbers into a Narrative**

```python
# Importing Libraries
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
iris=load_iris()
df = pd.DataFrame(iris.data,columns=iris.feature_names)

# Add species column
df["species"] = iris.target

# Convert numbers to species names
df["species"] = df["species"].map({
    0: "Setosa",
    1: "Versicolor",
    2: "Virginica"
})

# Display first 5 rows
print(df.head())
```

# **Average Petal Length**
  Which Iris species has the largest
  average petal length?

```python
# Calculate Average Petal Length
avg_petal = df.groupby("species")["petal length (cm)"].mean()
print(avg_petal)

# Creating Bar Chart

plt.figure(figsize=(8, 5))
plt.bar(
    avg_petal.index,
    avg_petal.values,
    color=["skyblue", "orange", "green"]
)
plt.title("Average Petal Length by Iris Species")
plt.xlabel("Iris Species")
plt.ylabel("Average Petal Length (cm)")
plt.show()
```

# Virginica has the largest average petal length, while Setosa has the smallest.

# **Average Sepal Length**

Which species has the longest average sepal?

```python
# Calculate Average Sepal Length
avg_sepal = df.groupby("species")["sepal length (cm)"].mean()

# Creating Bar Chart
plt.figure(figsize=(8, 5))
plt.bar(
    avg_sepal.index,
    avg_sepal.values,
    color=["skyblue", "orange", "green"]
)
plt.title("Average Sepal Length by Iris Species")
plt.xlabel("Iris Species")
plt.ylabel("Average Sepal Length (cm)")
plt.show()
```

# Virginica has the longest average sepal length, while Setosa has the shortest.

# **Species Distribution**
How many Iris samples belong to each species?

```python
# Count Species
species_counts = df["species"].value_counts()

print(species_counts)

# Creating Pie Chart
plt.figure(figsize=(7, 7))
plt.pie(
    species_counts.values,
    labels=species_counts.index,
    autopct="%1.1f%%",
    colors=["skyblue", "orange", "green"]
)
plt.title("Distribution of Iris Samples by Species")
plt.show()
```

# The dataset is evenly balanced, with each species representing one-third of the samples.

# **Petal Length Across Samples**
 Petal Length Across Samples


```python
# Creating Line Chart

plt.plot(df.index, df["petal length (cm)"])
plt.title("Petal Length Across Iris Samples")
plt.xlabel("Sample Number")
plt.ylabel("Petal Length (cm)")
plt.show()
```

# Petal length varies across the Iris samples, showing that different flowers have different petal sizes.
